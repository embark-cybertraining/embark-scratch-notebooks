```python
# Part B setup continued
import xarray as xr
```

```python
%%time
 # how much time does this step take?

# Part B Step 1 Find
doppio_url = 'https://tds.marine.rutgers.edu/thredds/dodsC/roms/doppio/2017_da/his/History_Best'
ds_roms = xr.open_dataset(doppio_url)
ds_roms

# Sage's output rendered <xarray.Dataset> Size: 7TB why not here?
```

```python
# Part B Step 2 Access (Parse and Load)

# Parse requires 'eta_idx' and 'xi_idx'
# Selected grid point: eta=41, xi=39
eta_idx = 41
xi_idx = 39
```

```python
%%time
 # how much time does this step take?


# Load Selected ROMS Profiles
ds_profile_temp = ds_roms['temp'].sel(time=slice('2026-03-01', '2026-04-01'), eta_rho=eta_idx, xi_rho=xi_idx).load()
ds_profile_sal = ds_roms['salt'].sel(time=slice('2026-03-01', '2026-04-01'), eta_rho=eta_idx, xi_rho=xi_idx).load()
```
